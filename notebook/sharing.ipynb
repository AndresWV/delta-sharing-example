```python
import os
import delta_sharing

profile = os.path.abspath("../config/open-datasets.share")
if not os.path.isfile(profile):
    raise FileNotFoundError(f"Delta Sharing profile not found: {profile}")

open_client = delta_sharing.SharingClient(profile)
open_client.list_all_tables()
```

```python
table_url = profile + "#delta_sharing.default.COVID_19_NYT"
data = delta_sharing.load_as_pandas(table_url, limit=10)
```
